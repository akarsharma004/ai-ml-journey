```python
# libraries for word embedding
# For one hot encoder
from tensorflow.keras.preprocessing.text import one_hot
```

```python
# sentences
sent =['the glass of milk',
       'the glass of juice',
       'She sup of teal' ,
       'I am a good boy' ,
       'I am a good developer' ,
       'understand the meaning of words',
       'your videos are good' ,]
```

```python
# vocabulary size
voc_size = 10000
```

```python
# One Hot Representation
one_hot_repr=[one_hot(words,voc_size) for words in sent]
one_hot_repr
```

```python
# word embedding representation
from tensorflow.keras.layers import Embedding, Input
from tensorflow.keras.utils import pad_sequences # used to cover the missing words to make each sentence with same words with 0's
from tensorflow.keras.models import Sequential
```

```python
# Padding sequencing sentences
sent_length = 8
embedded_docs = pad_sequences(one_hot_repr, padding='pre',maxlen=sent_length)
```

```python
embedded_docs
```

```python
# feature representation
dim_len = 10
```

```python
# model wrt to embedding layer
model = Sequential()

model = Sequential([
    Input(shape=(sent_length,)),
    Embedding(voc_size, dim_len)
])

model.summary()

```

```python
model.predict(embedded_docs)
```
