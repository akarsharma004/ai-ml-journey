### Prediction on trained ANN model

```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import pickle
import pandas as pd
import numpy as np
```

```python
# loading ann trained model , scaler, label encoder & ohe pickle
model = load_model('model.h5')
with open('ohe_geography.pkl','rb') as file:       
    ohe_geo = pickle.load(file)       
with open('label_encoder_gender.pkl','rb') as file:       
    label_encoder_gender = pickle.load(file)       
with open('scaler.pkl','rb') as file:       
    scaler = pickle.load(file)       
```

```python
# Example input data for prediction
input_data = {
    'CreditScore':600,
    'Geography': 'France',
    'Gender':'Male',
    'Age':40,
    'Tenure':3,
    'Balance':60000,
    'NumOfProducts':2,
    'HasCrCard':1,
    'IsActiveMember':1,
    'EstimatedSalary':50000
}
input_df=pd.DataFrame([input_data])
```

```python
# OHE the input feature
geo_ohe = ohe_geo.transform([[input_data['Geography']]]).toarray() # giving as list of featre
geo_ohe_df = pd.DataFrame(geo_ohe,columns=ohe_geo.get_feature_names_out(['Geography']))
```

```python
# Encoding input gender
input_df['Gender']=label_encoder_gender.transform(input_df['Gender'])
```

```python
# ohe column with input data
input_df = pd.concat([input_df.drop('Geography',axis=1),geo_ohe_df],axis=1)
```

```python
# scaling the input data
input_scaled = scaler.transform(input_df)
```

```python
# Prediction
prediction=model.predict(input_scaled)
prediction
```

```python
pp =prediction[0][0] # prediction probability
```

```python
if pp > 0.5:
    print('The customer is likely to churn.')
else:
    print('The cusomer is not likely to churn.')
```
