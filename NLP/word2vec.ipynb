## Model Implementation

```python
import pandas as pd
import re
import numpy as np
messages  = pd.read_csv("Data\spam.csv",sep =',', names =["label","message"], skiprows = 1)
```

```python
# Data preprocessing
from nltk.stem import WordNetLemmatizer
wnl = WordNetLemmatizer()
corpus = []
for i in range(0,len(messages)):
    review = re.sub('[^a-zA-Z]',' ',messages["message"][i])
    review = review.lower()
    review = review.split()
    review = [wnl.lemmatize(word) for word in review]
    review = ' '.join(review)
    corpus.append(review)
```

```python
from nltk import sent_tokenize
from gensim.utils import simple_preprocess
```

```python
# each sentences with there seperate words , lower casing 
words = []
for sentences in corpus:
    sent_token = sent_tokenize(sentences)
    for sent in sent_token:
        words.append(simple_preprocess(sent))
```

```python
import gensim
#training the word2vec model
model=gensim.models.Word2Vec(words)
```

```python
# to get all the vocabulary
model.wv.index_to_key
```

```python
# total no. of vocab
model.corpus_count
```

```python
# epoch
model.epochs
```

```python
# applying average word2vec
def avg_word2vec(doc):

    valid_words = [
        model.wv[word]
        for word in doc
        if word in model.wv.index_to_key
    ]

    if len(valid_words) == 0:
        return np.zeros(100, dtype=np.float32)

    return np.mean(valid_words, axis=0)
```

```python
from tqdm import tqdm # for progression visual
```

```python
X= []
for i in tqdm(range(len(words))):
    X.append(avg_word2vec(words[i]))
```

```python
X[0]
```

```python
# dependent features
y = messages[list(map(lambda x: len(x)>0,corpus))]
y = pd.get_dummies(y['label'])
y = y.iloc[:,0].values
```

```python
# converting Word2Vec sentence embeddings into a DataFrame for model input
df = pd.DataFrame(X)
# df = df.append(pd.DataFrame(X[i].reshape(1,-1),ignore_index=True))
```

```python
df['Output'] = y
```

```python
df.dropna(axis=0,inplace=True)
```

```python
# Independent Feature
X =df
# dependent feature
y = df['Output']
```

```python
# Train test split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.20)
```

```python
# Model Import
from sklearn.ensemble import RandomForestClassifier
classifier = RandomForestClassifier()
```

```python
X_train.columns = X_train.columns.astype(str)
X_test.columns = X_test.columns.astype(str)
```

```python
classifier.fit(X_train,y_train)
```

```python
y_pred = classifier.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report
print(accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
```
